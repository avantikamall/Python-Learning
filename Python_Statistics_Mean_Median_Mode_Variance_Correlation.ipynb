Measure of Dispersion — Interview Answer

You can say:

“Measure of dispersion tells us how much the data values are spread or scattered around a central value, such as the mean. The commonly used measures of dispersion are Range, Variance, Standard Deviation, and Interquartile Range (IQR).”

If the interviewer asks “Why do we use it?”, say:

“We use measures of dispersion to understand the variability or consistency of the data. Two datasets can have the same mean but different levels of spread.”

Simple example

Salary A: 40K, 41K, 40K, 39K, 40K
Salary B: 20K, 60K, 40K, 70K, 10K

Both may have a similar average, but Salary B has much higher dispersion because the values are more spread out.

One-line interview definition:

“Dispersion measures how widely the data points are distributed around the central tendency.”

1. Mean

“Mean is the average of all the values. We calculate it by adding all the values and dividing the sum by the number of values.”

Example: 10, 20, 30

Mean = (10 + 20 + 30) / 3 = 20

2. Median

“Median is the middle value when the data is arranged in ascending or descending order. If there is an even number of values, we take the average of the two middle values.”

Example: 10, 20, 30, 40, 50

Median = 30

3. Mode

“Mode is the value that occurs most frequently in the dataset.”

Example: 10, 20, 20, 30, 40

Mode = 20

⭐ Interview-ready short answer

“Mean represents the average, Median represents the middle value, and Mode represents the most frequently occurring value. Together, they are called measures of central tendency.”

Covariance and correlation both measure the relationship between two variables. Covariance tells us the direction of the relationship, whether it is positive or negative, but its value depends on the units of the variables. Correlation is a standardized form of covariance, with a value between -1 and +1, so it tells us both the direction and strength of the linear relationship.”

Simple example

Suppose:

days = [1, 2, 3, 4, 5]
weight = [90, 80, 70, 60, 50]

As days increase, weight decreases.

So:

Covariance → Negative
Correlation → Close to -1 because the relationship is strongly linear.
Main difference
	Covariance	Correlation
Tells direction?	✅ Yes	✅ Yes
Tells strength?	Not directly	✅ Yes
Range	No fixed range	-1 to +1
Units	Has units	Unit-free
Easy to interpret	Less	More
One line to remember

Covariance tells us the direction; correlation tells us the direction and strength.

And since you're learning Python:

import numpy as np

days = [1, 2, 3, 4, 5]
weight = [90, 80, 70, 60, 50]

covariance = np.cov(days, weight)[0, 1]
correlation = np.corrcoef(days, weight)[0, 1]

print("Covariance =", covariance)
print("Correlation =", correlation)

Interview tip: Don't say covariance always tells the strength. Its magnitude is affected by the variables' scales/units, which is one reason correlation is easier to compare across datasets.
