```python
# w a  function to take a price and qty from the user then calculate the total amount

#function cal bill dis 5% on tal amount then cal the total piad amount ta-d





def calBill(p,q):

  ta=p*q

  dis=ta*5/100

  tp=ta-dis

  return ta,dis,tp

price=int(input("Enter the price"))



qty=int(input("Enter the qty"))



ta,d,tp=calBill(price,qty)

print("Total Amount", ta)

print("Disscount",d)

print("Total Paid ",tp)
```

```python
#  w a p take list in finction then show itis values ,# W a function take list show odd numbers

def show(L):

  for i in range(0,len(L)):
    if(L[i]%2==0):
     print(L[i],"Even")
    else:
      print(L[i],"odd")




sales=[33,56,78,99,34,56,78,90,45]
show(sales)



```

```python
# w a function takes marks of list then count how many students are passed

def fun(Marks):

  count=0

  for i in range(0,len(Marks),1):

    if(Marks[i]>=40):

      count=count+1

  return count
marks=[45,67,89,45,6,7,89,34,56]

c=fun(marks)

print(c)
```
