### College Example: Student Grades with Pandas Series

Let's imagine we want to keep track of a student's grades for different courses. We can use a `pandas.Series` where the index represents the course names and the values represent the grades.

```python
student_grades = pd.Series(
    [85, 92, 78, 88, 95],
    index=['Calculus I', 'Physics I', 'English Literature', 'Computer Science I', 'Philosophy']
)

print("Alice Johnson's Grades:")
print(student_grades)


average_grade = student_grades.mean()
print(f"\nAlice Johnson's Average Grade: {average_grade:.2f}")

# Find grades above 90
good_grades = student_grades[student_grades > 90]
print("\nCourses with grades above 90:")
print(good_grades)
```

### Simple Example: Grocery List and Costs

Let's say you have a grocery list with items and their respective prices. We can represent this with two `pd.Series`: one for the prices and one for the quantities, and then calculate the total cost.

```python
item_prices = pd.Series(
    [2.50, 3.00, 1.25, 4.75, 0.99],
    index=['Apples', 'Milk', 'Bread', 'Chicken', 'Eggs']
)

item_quantities = pd.Series(
    [4, 1, 2, 1, 12],
    index=['Apples', 'Milk', 'Bread', 'Chicken', 'Eggs']
)

print("Item Prices:")
print(item_prices)
print("\nItem Quantities:")
print(item_quantities)

cost_per_item = item_prices * item_quantities
print("\nCost per Item:")
print(cost_per_item)

# Calculate the total grocery bill
total_bill = cost_per_item.sum()
print(f"\nTotal Grocery Bill: ${total_bill:.2f}")
```
