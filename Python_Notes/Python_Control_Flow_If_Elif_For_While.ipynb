```python
print("1.. Perimeter of Rectangle")
print("2.. Circumference of Circle")
print("3.. Volume of Cube")
print("4.. Volume of Cuboid")
print("5.. Exit")

ch = int(input("Enter Your Choice No.: 1-5: "))

if ch == 1:
    Num1 = int(input("Enter Length: "))
    Num2 = int(input("Enter Width: "))

    Perimeter_of_Rectangle = 2 * (Num1 + Num2)
    print("Perimeter of Rectangle:", Perimeter_of_Rectangle)

elif ch == 2:
    Num1 = int(input("Enter Radius: "))

    pie = 3.14
    Circumference_of_Circle = 2 * pie * Num1
    print("Circumference of Circle:", Circumference_of_Circle)

elif ch == 3:
    Num1 = int(input("Enter Side: "))

    Volume_of_Cube = Num1 ** 3
    print("Volume of Cube:", Volume_of_Cube)

elif ch == 4:
    Num1 = int(input("Enter Length: "))
    Num2 = int(input("Enter Width: "))
    Num3 = int(input("Enter Height: "))

    Volume_of_Cuboid = Num1 * Num2 * Num3
    print("Volume of Cuboid:", Volume_of_Cuboid)

elif ch == 5:
    print("Exit")

else:
    print("Invalid Choice")
```
