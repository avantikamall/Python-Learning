```python
import matplotlib.pyplot as plt
Days=["Sun","Mon","Tues","Wed","Thrus","Fri","Sat"]
Temp=[28,30,42,45,50,2,7]
plt.plot(Days,Temp,marker='*')
plt.title("Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
Books=["IT Books","Non IT Books"]
Sales=[2000,4000]
color=["red","green"]
plt.bar(Books,Sales,color=color)
plt.title("Books Sales")
plt.xlabel("Books")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
Books=["IT Books","Non IT Books"]
Sales=[2000,4000]
color=["red","green"]
plt.barh(Books,Sales,color=color)
plt.title("Books Sales")
plt.xlabel("Books")
plt.ylabel("Sales")
plt.grid()
plt.show()
```
