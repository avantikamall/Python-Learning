
"""

Perform the following tasks:

1. Add `"jatin"` at the end of the list using `append()`.

2. Add `"rohit"` and `"neha"` to the list using `extend()`.

3. Insert `"sachin"` at index `2` using `insert()`.

4. Remove the last user using `pop()`.

5. Remove the user at index `1` using `pop(1)`.

6. Sort the users in **ascending order** using `sort()`.

7. Sort the users in **descending order** using `sort(reverse=True)`.

8. Count how many times `"ajay"` appears using `count()`.

9. Print the final list.

"""

```python
users=["ajay","asha","deepa","kavita","jatin","sachin"]
#Add "jatin" at the end of the list using append().
users.append("Jatin")
# Add "rohit" and "neha" to the list using extend().
users.extend(["rohit","neha"])
# Insert "sachin" at index 2 using insert().
users.insert(2,"sachin")
#Remove the last user using pop().
users.pop()
#Remove the user at index 1 using pop(1).
users.pop(1)
#Sort the users in ascending order using sort().
users.sort()
#Sort the users in descending order using sort(reverse=True).
users.sort(reverse=True)
#Count how many times "ajay" appears using count().
users.count("ajay")
#Print the final list.
print(users)
```
